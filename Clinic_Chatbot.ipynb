Setup Clinic Rules


```python
from datetime import datetime, timedelta
from zoneinfo import ZoneInfo

TZ = ZoneInfo("Asia/Karachi")
OPEN_HOUR, OPEN_MIN = 17, 30      # 5:30 PM
SLOT_MINUTES = 45
SLOTS_PER_DAY = 5
MIN_DAYS_AHEAD = 3
CLOSED_WEEKDAY = 6                # Sunday (Monday = 0)
```

Each Day 5 slots

```python
def generate_day_slots(day):
    start = datetime(day.year, day.month, day.day, OPEN_HOUR, OPEN_MIN, tzinfo=TZ)
    slots = []
    for i in range(SLOTS_PER_DAY):
        s = start + timedelta(minutes=SLOT_MINUTES * i)
        e = s + timedelta(minutes=SLOT_MINUTES)
        slots.append((s, e))
    return slots

# test
from datetime import date
for s, e in generate_day_slots(date(2026, 11, 7)):
    print(s.strftime("%I:%M %p"), "->", e.strftime("%I:%M %p"))
```

Remove Busy Scheducle from Calender

```python
def get_busy_times(day):
    day_start = datetime(day.year, day.month, day.day, 0, 0, tzinfo=TZ)
    day_end = day_start + timedelta(days=1)
    body = {
        "timeMin": day_start.isoformat(),
        "timeMax": day_end.isoformat(),
        "timeZone": "Asia/Karachi",
        "items": [{"id": "primary"}],
    }
    result = calendar_service.freebusy().query(body=body).execute()
    busy = result["calendars"]["primary"]["busy"]
    return [(datetime.fromisoformat(b["start"]).astimezone(TZ),
             datetime.fromisoformat(b["end"]).astimezone(TZ)) for b in busy]
```

Implement Rules and Fetch Free Slots

```python
def get_available_slots(day, emergency=False):
    now = datetime.now(TZ)
    today = now.date()

    if day.weekday() == CLOSED_WEEKDAY:
        return [], "Clinic is closed on Sunday."

    if emergency:
        if day != today:
            return [], "Emergency appointments are same-day only."
    else:
        earliest = today + timedelta(days=MIN_DAYS_AHEAD)
        if day < earliest:
            return [], f"Bookings need at least {MIN_DAYS_AHEAD} days notice. Earliest date: {earliest}."

    busy = get_busy_times(day)
    free = []
    for s, e in generate_day_slots(day):
        if emergency and s <= now:
            continue
        if any(s < b_end and e > b_start for b_start, b_end in busy):
            continue
        free.append(s)

    return (free, "OK") if free else ([], "No slots available that day.")
```

Making event automatically


```python
def create_appointment_event(name, email, service, start, emergency=False):
    end = start + timedelta(minutes=SLOT_MINUTES)

    # Double-booking se bachao: banane se pehle dobara check
    free, _ = get_available_slots(start.date(), emergency=emergency)
    if start not in free:
        return None, "Sorry, that slot is no longer available."

    event = {
        "summary": f"{service} - {name}",
        "description": f"Patient: {name}\nEmail: {email}\nService: {service}\nBooked via AI assistant",
        "start": {"dateTime": start.isoformat(), "timeZone": "Asia/Karachi"},
        "end":   {"dateTime": end.isoformat(),   "timeZone": "Asia/Karachi"},
        "attendees": [{"email": email}],
    }
    created = calendar_service.events().insert(
        calendarId="primary", body=event, sendUpdates="none"
    ).execute()
    return created, "Booked"
```

Divide PDF in chunks with section wise

```python
import re

def smart_chunks(text, max_chars=500):
    sections = re.split(r"\n(?=\d+\.\s[A-Z])", text)   # "1. Clinic...", "2. ..." par todo
    chunks = []
    for sec in sections:
        lines = [l.strip() for l in sec.split("\n") if l.strip()]
        if not lines:
            continue
        title = lines[0]
        current = title
        for line in lines[1:]:
            if len(current) + len(line) + 1 > max_chars:
                chunks.append(current)
                current = title + "\n" + line      # naye tukre mein bhi title
            else:
                current += "\n" + line
        chunks.append(current)
    return chunks

chunks = smart_chunks(text)
print("Total chunks:", len(chunks))
for c in chunks[:3]:
    print("-----")
    print(c)
```

Embeddings

```python
from sentence_transformers import SentenceTransformer
import numpy as np

embedder = SentenceTransformer("all-MiniLM-L6-v2")
chunk_vectors = embedder.encode(chunks, normalize_embeddings=True)
print(chunk_vectors.shape)
```

Search Function


```python
def search_clinic_info(question, top_k=3):
    q = embedder.encode([question], normalize_embeddings=True)[0]
    scores = chunk_vectors @ q
    best = np.argsort(scores)[::-1][:top_k]
    return "\n\n---\n\n".join(chunks[i] for i in best)
```

```python
import re, json

SERVICES = [
    "Consultation", "Dental X-ray", "Scaling and Polishing", "Tooth Filling",
    "Root Canal", "Tooth Extraction", "Wisdom Tooth Removal", "Dental Crown",
    "Denture", "Dental Implant", "Teeth Whitening", "Veneers",
    "Braces", "Clear Aligners",
]
EMAIL_RE = re.compile(r"^[^@\s]+@[^@\s]+\.[^@\s]+$")

def parse_day(date_str):
    return datetime.strptime(date_str, "%Y-%m-%d").date()

def tool_search_clinic_info(question):
    return search_clinic_info(question, top_k=3)

def tool_get_available_slots(date, emergency=False):
    try:
        day = parse_day(date)
    except ValueError:
        return json.dumps({"error": "Date must be in YYYY-MM-DD format."})
    slots, msg = get_available_slots(day, emergency=emergency)
    return json.dumps({"date": date, "message": msg,
                       "slots": [s.strftime("%H:%M") for s in slots]})

def tool_book_appointment(name, email, service, date, time,
                          user_confirmed=False, emergency=False):
    if not user_confirmed:
        return json.dumps({"error": "Not booked. Show the user a summary and ask them to confirm first."})
    if not name or not name.strip():
        return json.dumps({"error": "Name is required."})
    if not EMAIL_RE.match(email or ""):
        return json.dumps({"error": "Invalid email address."})
    if service not in SERVICES:
        return json.dumps({"error": f"Unknown service. Choose from: {SERVICES}"})
    try:
        day = parse_day(date)
        t = datetime.strptime(time, "%H:%M").time()
    except ValueError:
        return json.dumps({"error": "Use date YYYY-MM-DD and time HH:MM (24-hour)."})

    start = datetime.combine(day, t, tzinfo=TZ)
    event, status = book_appointment(name.strip(), email.strip(), service, start, emergency)
    if not event:
        return json.dumps({"booked": False, "message": status})
    return json.dumps({"booked": True, "message": status,
                       "date": date, "time": time, "service": service})
```

```python
TOOLS = [
    {"type": "function", "function": {
        "name": "tool_search_clinic_info",
        "description": "Search clinic knowledge base: services, prices, timings, policies, FAQs, contact info. Use for ANY question about the clinic.",
        "parameters": {"type": "object",
            "properties": {"question": {"type": "string"}},
            "required": ["question"]}}},
    {"type": "function", "function": {
        "name": "tool_get_available_slots",
        "description": "Get free appointment slots for a date. Set emergency=true only for a same-day dental emergency.",
        "parameters": {"type": "object",
            "properties": {
                "date": {"type": "string", "description": "YYYY-MM-DD"},
                "emergency": {"type": "boolean"}},
            "required": ["date"]}}},
    {"type": "function", "function": {
        "name": "tool_book_appointment",
        "description": "Book an appointment. Call ONLY after the user has chosen a slot and explicitly confirmed the summary.",
        "parameters": {"type": "object",
            "properties": {
                "name": {"type": "string"},
                "email": {"type": "string"},
                "service": {"type": "string", "enum": SERVICES},
                "date": {"type": "string", "description": "YYYY-MM-DD"},
                "time": {"type": "string", "description": "24-hour HH:MM, must be one of the slots returned"},
                "user_confirmed": {"type": "boolean", "description": "true only if the user said yes to the final summary"},
                "emergency": {"type": "boolean"}},
            "required": ["name", "email", "service", "date", "time", "user_confirmed"]}}},
]

TOOL_FUNCTIONS = {
    "tool_search_clinic_info": tool_search_clinic_info,
    "tool_get_available_slots": tool_get_available_slots,
    "tool_book_appointment": tool_book_appointment,
}

def run_tool(name, args):
    try:
        return TOOL_FUNCTIONS[name](**args)
    except Exception as e:
        return json.dumps({"error": f"Tool failed: {e}"})
```

Booking Flow

```python
YES = {"yes", "yeah", "yep", "y", "confirm", "ok", "okay", "sure", "haan", "han", "ji", "bilkul"}
NO  = {"no", "nahi", "nahin", "cancel", "wait", "stop"}
def is_yes(text):
    w = set(re.findall(r"[a-z]+", text.lower()))
    return bool(w & YES) and not (w & NO)

def new_state():
    return {"stage": "menu", "service": None, "name": None, "email": None,
            "date": None, "time": None, "emergency": False, "date_map": {}, "slot_map": {}}

def fmt_time(hhmm):
    return datetime.strptime(hhmm, "%H:%M").strftime("%I:%M %p").lstrip("0")

def match_service(text):
    for s in SERVICES:
        if s.lower() in text.lower():
            return s
    return None

def enter_pick_date(st, n=6):
    d = datetime.now(TZ).date() + timedelta(days=MIN_DAYS_AHEAD)
    dates = []
    while len(dates) < n:
        if d.weekday() != CLOSED_WEEKDAY:
            dates.append(d)
        d += timedelta(days=1)
    st["date_map"] = {x.strftime("%a %d %b"): x.isoformat() for x in dates}
    st["stage"] = "pick_date"

def show_slots(st, day):
    slots, msg = get_available_slots(day, emergency=st["emergency"])
    if not slots:
        was_emergency = st["emergency"]
        st["emergency"] = False
        enter_pick_date(st)
        if was_emergency:
            return "No emergency slots are left today. Please call 0333-6113301, or choose a normal date:"
        return f"{msg} Please choose another date:"
    st["date"] = day.isoformat()
    st["slot_map"] = {fmt_time(s.strftime("%H:%M")): s.strftime("%H:%M") for s in slots}
    st["stage"] = "pick_slot"
    return f"Available times on {day.strftime('%A, %d %B %Y')}. Please choose one:"

def begin_booking(st):
    if st["service"]:
        st["stage"] = "ask_name"
        return f"Sure! Let's book your {st['service']} appointment.\n\n1) What's your name?"
    st["stage"] = "ask_service"
    return "Sure! Which service would you like to book?"

def free_text(text, chat, st):
    if text in SERVICES:                                   # service button dabaya
        st["service"], st["stage"] = text, "service_info"
        answer, _ = rag_answer(f"What is the price of {text}? Mention how long the visit takes.", [])
        return f"Okay! {answer}\n\nWould you like to book an appointment for this service?"
    history = [{"role": m["role"], "content": m["content"]} for m in chat[-7:-1]]
    answer, call = rag_answer(text, history)
    if call is not None or BOOK_RE.search(text):
        call = call or {}
        s = match_service(str(call.get("service") or "")) or match_service(text)
        if s:
            st["service"] = s
        st["emergency"] = bool(call.get("emergency")) or bool(EMERG_RE.search(text))
        return begin_booking(st)
    return answer

def process(text, chat, st):
    text = text.strip()
    chat = chat + [{"role": "user", "content": text}]
    stage = st["stage"]
    try:
        if text in ("Cancel booking", "Back to services", "See other services"):
            st = new_state()
            reply = "No problem! Choose a service below, or ask me anything."
        elif text == "Want to book appointment":
            reply = begin_booking(st)
        elif stage == "ask_service":
            s = match_service(text)
            if s:
                st["service"], st["stage"] = s, "ask_name"
                reply = f"Great, {s}.\n\n1) What's your name?"
            else:
                reply = "Please pick a service from the buttons below."
        elif stage == "ask_name":
            if 2 <= len(text) <= 60:
                st["name"], st["stage"] = text, "ask_email"
                reply = f"Nice to meet you, {text}!\n\n2) Can you tell me your email address?"
            else:
                reply = "Please type your full name."
        elif stage == "ask_email":
            if EMAIL_RE.match(text):
                st["email"] = text
                if st["emergency"]:
                    reply = "Thank you. Since this is an emergency, here are today's slots.\n\n" + show_slots(st, datetime.now(TZ).date())
                else:
                    enter_pick_date(st)
                    reply = ("Thank you! Please choose a date. Bookings need at least 3 days' notice and we are closed on Sundays. "
                             "For a dental emergency, press Emergency (today).")
            else:
                reply = "That email doesn't look right. Please type it again (for example name@gmail.com)."
        elif stage == "pick_date":
            if text == "Emergency (today)":
                st["emergency"] = True
                reply = show_slots(st, datetime.now(TZ).date())
            elif text in st["date_map"]:
                st["emergency"] = False
                reply = show_slots(st, datetime.fromisoformat(st["date_map"][text]).date())
            else:
                reply = "Please choose a date from the buttons."
        elif stage == "pick_slot":
            if text == "Pick another date":
                enter_pick_date(st)
                reply = "Sure, choose another date:"
            elif text in st["slot_map"]:
                st["time"], st["stage"] = st["slot_map"][text], "confirm"
                day = datetime.fromisoformat(st["date"]).strftime("%A, %d %B %Y")
                reply = (f"Please check your booking:\n\n- Name: {st['name']}\n- Email: {st['email']}\n"
                         f"- Service: {st['service']}\n- Date: {day}\n- Time: {text}\n\nShall I confirm this booking?")
            else:
                reply = "Please choose a time from the buttons."
        elif stage == "confirm":
            if text == "Yes, confirm" or is_yes(text):
                res = json.loads(tool_book_appointment(
                    st["name"], st["email"], st["service"], st["date"], st["time"],
                    user_confirmed=True, emergency=st["emergency"]))
                if res.get("booked"):
                    st["stage"] = "done"
                    reply = (f"Your appointment is booked! A confirmation email has been sent to {st['email']}. "
                             "See you at the clinic!")
                    if "email failed" in res.get("message", ""):
                        reply = "Your appointment is booked, but the confirmation email could not be sent."
                else:
                    enter_pick_date(st)
                    reply = f"Sorry, I couldn't book that: {res.get('message') or res.get('error')}\n\nPlease choose another date:"
            elif text == "No, change" or text.lower() in NO:
                enter_pick_date(st)
                reply = "No problem. Please choose a date:"
            else:
                reply = "Please press Yes, confirm or No, change."
        else:
            reply = free_text(text, chat, st)
    except Exception as e:
        reply = f"Sorry, something went wrong ({type(e).__name__}). Please try again."
    chat.append({"role": "assistant", "content": reply})
    return chat, st
```
